```python
# Imports
import pandas as pd
import numpy as np
import scvelo as scv
import anndata as adata
import scanpy as sc
```

```python
# Data
ann_data = adata.read_h5ad("C:/Users/joshu/Desktop/RNAVelo_Tools_Comparison/data/janssens_2025_CellRanger_Velocyto_counts_eLNP.h5ad")
ann_data
```

```python
# Preprocessing
scv.pp.filter_and_normalize(
    ann_data,
    min_shared_counts=20,
)

scv.pp.moments(
    ann_data, 
    n_pcs = 30,
    n_neighbors = 30 
)

sc.tl.umap(ann_data)
```

```python
# Recover the splicing kinetics of the genes

scv.tl.recover_dynamics(ann_data, n_jobs = 4, n_top_genes = 2000)
```

```python
# Find the dynamical RNA Velocity

scv.tl.velocity(
    ann_data,
    mode = "dynamical"
)
```

```python
# Create the RNA Velocity graph
scv.pl.velocity_embedding_stream(
    ann_data,
    basis="umap",
    color="newClustersTmp",
    legend_fontsize=9,
    dpi=150            # increase dpi for higher resolution
)
```
