```python
# Preprocess the Data
scv.pp.filter_and_normalize(
    adata, 
    min_shared_counts = 20, 
    n_top_genes = 2000 # Filters for the top 2000 genes
)
scv.pp.moments(
    adata, 
    n_neighbors = 30, # Looks at the 30 nearest neighbors
    n_pcs = 30 # Looks at the 30 principle components
)
```

```python
# Train the model
trainer = dv.train(
    adata, 
    dv.Constants.default_configs
)
```

```python
# Crate the umap of the clusters
sc.pp.neighbors(
    adata,
    n_neighbors=30,
    n_pcs=30
)

sc.tl.umap(adata)
```

```python
# Create the velocity graph
scv.tl.velocity_graph(
    adata,
    n_jobs=4
)
```

```python
# Plot the velocity results
scv.pl.velocity_embedding_stream(
    adata,
    basis = "umap",
    color = "newClustersTmp",
    legend_fontsize = 9,
    dpi = 150
)
```
